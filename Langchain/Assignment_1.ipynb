## Task 1 : Replace Hardcoded Prompts
### Converting the following hardcoded prompt function into a Langchain Prompt Template.
```python
def explain_topic(topic):
   return f"Explain {topic} in simple terms for beginners"
```

````python
# Using PromptTemplate as it is used for Simple string based template (used for normal LLMs)
from langchain_core.prompts import PromptTemplate
prompt_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)
def explain_topic(topic):
    prompt = prompt_template.format(topic=topic)
    print(prompt)
while True:
    user_input = input("Enter the topic : ")
    if user_input=="quit":
        break
    explain_topic(user_input)
````

## Task 2: Multi-Input Prompt System

````python
prompt_template = PromptTemplate(
    input_variables=["topic","audience","tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)
while True:
    user_choice = input("Want to quit : ")
    if user_choice == "yes":
        break
    user_topic = input("Enter the topic : ")
    user_audience = input("Enter the audience : ")
    user_tone = input("Enter the tone : ")
    prompt = prompt_template.format(topic=user_topic,audience=user_audience,tone=user_tone)
    print(prompt)
````

## Task 3: Prompt Variations Engine

````python
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)
interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 questions about {topic}"
)
storytelling_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as an engaging story"
)
while True:
    user_choice = input("Want to quit : ")
    if user_choice == "yes":
        break
    user_choice = input("Enter the choice (teaching | interview | storytelling) : ")
    match user_choice:
        case "teaching":
            user_topic = input("Enter the topic : ")
            prompt = teaching_template.format(topic=user_topic)
            print(prompt)
        case "interview":
            user_topic = input("Enter the topic : ")
            prompt = interview_template.format(topic=user_topic)
            print(prompt)
        case "storytelling":
            user_topic = input("Enter the topic : ")
            prompt = storytelling_template.format(topic=user_topic)
            print(prompt)
        case _ :
            print("Invalid category entered!")
            break
````

## Task 4 : ChatPrompt Template System

````python
from langchain_core.prompts import ChatPromptTemplate
teacher_template = ChatPromptTemplate.from_messages(
    [
        ("system","Act as an expert Engineering Teacher and answer the following user question"),
        ("user","Explain {topic} in a detailed way")
    ]
)
interviewer_template = ChatPromptTemplate.from_messages(
    [
        ("system","You are an Interviewer for the Candidate(user) answer user's questions in a professional way . Also ask hard follow up questions . No sugar of Responses go straight forward , precise , accurate , complete and in a Responsible way!"),
        ("user","Hello , My name is {username} and I have applied for {user_role} role")
    ]
)
motivator_template = ChatPromptTemplate.from_messages(
    [
        ("system","You are an Expert MOTIVATOR your client is user , go polite and positive tone , help user to get motivated"),
        ("user","Hello , I'm suffering from {user_problem} problem please can you help me to get out of this ?")
    ]
)
while True:
    user_choice = input("Want to quit : ")
    if user_choice == "yes":
        break
    user_choice = input("Enter the choice (teacher | interviewer | motivator) : ")
    match user_choice:
        case "teacher":
            user_topic = input("Enter the topic : ")
            prompt = teacher_template.format_messages(topic=user_topic)
            print(prompt)
            print(teacher_template.format(topic=user_topic))
            print("\n")
        case "interviewer":
            user_name = input("Enter your Name : ")
            user_role = input("Enter your Role : ")
            prompt = interviewer_template.format_messages(username=user_name,user_role=user_role)
            print(prompt)
            print(interviewer_template.format(username=user_name,user_role=user_role))
            print("\n")
        case "motivator":
            problem = input("Enter your Problem : ")
            prompt = motivator_template.format_messages(user_problem=problem)
            print(prompt)
            print(motivator_template.format(user_problem=problem))
            print("\n")
        case _ :
            print("Invalid category entered!")
            break
````

## Task 5: Input Validation Layer

````python
prompt_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

# Validation Function
def validate_inputs(audience, tone, strict=False): # by default keeping strict parameter as False so that it will select the default one
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    # Normalize input
    audience = audience.lower()
    tone = tone.lower()

    # Validate audience
    if audience not in valid_audience:
        if strict:
            raise ValueError(f"Invalid audience! Choose from {valid_audience}")
        else:
            print(" Invalid audience → defaulting to 'beginner'")
            audience = "beginner"

    # Validate tone
    if tone not in valid_tone:
        if strict:
            raise ValueError(f"Invalid tone! Choose from {valid_tone}")
        else:
            print(" Invalid tone → defaulting to 'formal'")
            tone = "formal"

    return audience, tone


while True:
    user_choice = input("Want to quit (yes/no): ")
    if user_choice.lower() == "yes":
        break

    user_topic = input("Enter the topic: ")
    user_audience = input("Enter the audience: ")
    user_tone = input("Enter the tone: ")

    try:
        # Validate inputs
        audience, tone = validate_inputs(user_audience, user_tone)

        # Generate prompt
        prompt = prompt_template.format(
            topic=user_topic,
            audience=audience,
            tone=tone
        )

        print("\nGenerated Prompt:")
        print(prompt)

    except ValueError as e:
        print(f" Error: {e}")
````

## Task 6 : Prompt Generator App

````python
prompt_template = PromptTemplate(
    input_variables=["topic", "audience", "tone", "style"],
    template="Explain {topic} for {audience} in a {tone} {style} style"
)

# Function to generate prompt
def generate_prompt(topic, audience, tone, style):
    valid_styles = ["teaching", "interview", "storytelling"]

    style = style.lower()

    if style not in valid_styles:
        print(" Invalid style → defaulting to 'teaching'")
        style = "teaching"

    prompt = prompt_template.format(
        topic=topic,
        audience=audience,
        tone=tone,
        style=style
    )

    return prompt



while True:
    user_choice = input("Want to quit : ")
    if user_choice == "yes":
        break

    user_topic = input("Enter the topic : ")
    user_audience = input("Enter the audience : ")
    user_tone = input("Enter the tone : ")
    user_style = input("Enter the style (teaching | interview | storytelling) : ")

    prompt = generate_prompt(
        topic=user_topic,
        audience=user_audience,
        tone=user_tone,
        style=user_style
    )

    print(prompt)
````

## Task 7: Template Reusability Test

````python
def prompt_pipeline():
    while True:
        user_choice = input("Want to quit : ")
        if user_choice == "yes":
            break

        # Validating User Inputs
        user_topic = input("Enter the topic : ")
        user_audience = input("Enter the audience : ")
        user_tone = input("Enter the tone : ")
        user_style = input("Enter the style (teaching | interview | storytelling) : ")

        user_audience, user_tone = validate_inputs(user_audience, user_tone)

        # Dynamic Prompt Generation (reuse existing function)
        prompt = generate_prompt(
            topic=user_topic,
            audience=user_audience,
            tone=user_tone,
            style=user_style
        )

        # Output
        print("Generated Prompt:")
        print(prompt)
prompt_pipeline()
````

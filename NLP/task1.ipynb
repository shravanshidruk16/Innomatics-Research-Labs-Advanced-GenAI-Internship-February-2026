## 2. Build Advanced Preprocessing Function

```python
import nltk , string , re
```

```python
from nltk.tokenize import word_tokenize
```

```python
def process_text(text):
    # 1. remove numbers
    text = "".join(char for char in text if char not in string.digits)
    
    # 2. removing extra spaces
    text = " ".join(text.split())
    
    # 3. handle repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    # removing repeated punctuations : one punctuation allowed
    # text = re.sub(r'[!?.]{2,}', '', text)
    # removing punctuations completely
    text = "".join(char for char in text if char not in string.punctuation)
    
    # 4. converting everything to lowercase
    text = text.lower()

    # 5. remove very short tokens (≤2) but keep "no", "not"
    tokens = word_tokenize(text)
    
    filtered_tokens = [
        word for word in tokens 
        if len(word) > 2 or word in ["no", "not"]
    ]
    
    # Join back
    text = " ".join(filtered_tokens)
    
    # 6. remove URLs , emails format
    text = re.sub(r'http\S+|www\S+', '', text)

    return text
```

## 3. Stress Testing

```python
def testing(sentences):
    counter = 1
    for sentence in sentences:
        print('\n')
        print(f"Sentence {counter}")
        text = process_text(sentence)
        print(f'Original Sentence : {sentence}')
        print(f'Cleaned Sentence : {text}')
        print(f'Cleaned Tokens : {word_tokenize(text)}')
        counter+=1
```

## 4. Token Analytics

```python
def token_analytics(sentences):
    counter = 1
    for sentence in sentences:
        text = process_text(sentence)
        tokens = word_tokenize(text)
        tokens_count = len(word_tokenize(text))
        unique_tokens_count = len(set(tokens))
        avg_token_length = sum(len(token) for token in tokens) / len(tokens)
        print('\n')
        print(f"Sentence {counter}")
        print(f'Original Sentence : {sentence}')
        print(f'Tokens created are : {tokens}')
        print(f'Total number of tokens : {tokens_count}')
        print(f'Total number of UNIQUE tokens : {unique_tokens_count}')
        print(f'Average token length for this sentence : {avg_token_length}')
        counter+=1
```

## 5. Frequency Analysis

```python
from collections import Counter
```

```python
def frequency_analysis(sentences):
    all_tokens = []
    
    # Preprocess + collect tokens
    for sentence in sentences:
        text = process_text(sentence)
        tokens = word_tokenize(text)
        all_tokens.extend(tokens)


    # Count frequencies
    freq = Counter(all_tokens)

    # Top 10 most frequent
    top_10 = freq.most_common(10)
    
    # Top 5 least frequent
    least_5 = freq.most_common()[:-6:-1]
    
    # Print results
    print("\nTop 10 Most Frequent Words:")
    for word, count in top_10:
        print(f"{word} : {count}")
    
    print("\nTop 5 Least Frequent Words:")
    for word, count in least_5:
        print(f"{word} : {count}")
    
    # return top_10, least_5
```

## 6. Building Full Pipeline

```python
def full_pipeline(text_list):
    clean_sentences = []
    all_tokens = []
    
    for text in text_list:
        # Step 1: Preprocess
        cleaned = process_text(text)
        clean_sentences.append(cleaned)
        
        # Step 2: Tokenize
        tokens = word_tokenize(cleaned)
        all_tokens.extend(tokens)
    
    # Step 3: Return structured output
    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```

```python
print(full_pipeline(test_input))
```

```python
def safe_full_pipeline(text_list):
    
    # Case 1: Empty input list
    if not text_list:
        return {
            "tokens": [],
            "clean_sentences": [],
            "error": "Input list is empty"
        }
    
    # Validate and clean input list
    valid_texts = []
    
    for text in text_list:
        
        # Case 2: Empty string or only spaces
        if not text or text.strip() == "":
            continue
        
        # Case 3: Only numbers
        if re.fullmatch(r'\d+', text.strip()):
            continue
        
        # Case 4: Only emojis / non-text characters
        # (No alphabets present)
        if not re.search(r'[a-zA-Z]', text):
            continue
        
        valid_texts.append(text)
    
    # If no valid text remains
    if not valid_texts:
        return {
            "tokens": [],
            "clean_sentences": [],
            "error": "No valid textual data found"
        }
    
    # Use your existing pipeline if everything is well
    result = full_pipeline(valid_texts)
    
    return result
```

```python
texts = [
    "",                      # empty
    "123456",                # only numbers
    "😂😂🔥🔥",            # only emojis
    "This is soooo good!!!", # valid
    "Visit http://test.com"
]

result1 = safe_full_pipeline(texts)
print(result1)
result2 = safe_full_pipeline(test_input)
print(result2)
```
